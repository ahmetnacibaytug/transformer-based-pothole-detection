# Pothole Detection
Run from the repository root after installing dependencies. Set paths below; datasets and checkpoints are not included. Training is optional.

Configure paths

```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd()
if not (ROOT / "src").is_dir(): ROOT = ROOT.parent
DATA = ROOT / "data/PotholeProjectv2"
WEIGHTS = ROOT / "weights/best_vit_pothole.pth"
```

Evaluate existing ViT checkpoint

```python
subprocess.run([sys.executable, str(ROOT/"src/classify.py"), "evaluate", "--model", "vit", "--data", str(DATA), "--weights", str(WEIGHTS), "--out", str(ROOT/"runs/vit")], check=True)
```
